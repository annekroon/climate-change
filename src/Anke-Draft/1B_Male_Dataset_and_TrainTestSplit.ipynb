# Split up a Train and Test Dataset of Tweets by Male Annotators

```python
import pandas as pd
import numpy as np
from scipy.stats import mode
```

## 1. Read and Modify Dataset

```python
# read annotations
df = pd.read_csv('../data/annotations/twitter_klimaat_discussie_coderingen.csv')
# merge in tweet texts
#tweets = pd.read_csv('../retrieve-tweets/klimaatgesprekken-eerste-mag-geen-at-bevatten-clean.csv')
tweets = pd.read_csv('../retrieve-tweets/klimaatgesprekken-clean.csv')
tweets = tweets.rename({'conversationid':"tweet_id", 'tweetorder':'comment_nr'}, axis=1)
df = df.merge(tweets, on=['tweet_id', 'comment_nr'], how='left')
# some data wrangling
# not considering incomplete rows
_l1 = len(df)
df.dropna(inplace=True)
print(f"Dropped {_l1-len(df)} of {_l1} rows due to missing values. New number of rows: {len(df)}")
df.rename(lambda x: x.replace('.','_'), axis=1, inplace=True)
df['tweet_id_comment_nr'] = df['tweet_id'].map(str) + "_" +  df['comment_nr'].map(str)
df.insert(list(df.columns).index('education'), 'higher_education', df.education.map(lambda x: int(x.startswith("Hoger") or x.startswith("Wetenschappelijk"))))

# recode textual labels to 0 and 1
#def customrecode(x):
#    if 'geen' in x.lower():
#        return False
#    elif 'oneens' in x.lower():
#        return False
#    else:
#        return True

#df.loc[:,[e for e in df.columns if e.startswith('comment_') and not e=='comment_nr']] = \
#    df.loc[:,[e for e in df.columns if e.startswith('comment_') and not e=='comment_nr']].map(customrecode)

df['comment_inhoudelijk_argument_binary'] = df['comment_inhoudelijk_argument'].map(lambda x: int(x=="Inhoudelijk argument"))
df['comment_persoonlijke_aanval_binary'] = df['comment_persoonlijke_aanval'].map(lambda x: int(x=="Persoonlijke aanval"))
df['comment_respectvol_binary'] = df['comment_respectvol'].map(lambda x: int(x=="Eens"))
```

## 2. Make A Male Subsamples

```python
df_male = df_red[df_red['gender'] == "Man"]
df_male
```

## 3. Pivot the Dataset and get majority votes

### 3.1 Respect

```python
# Step 1: Add a counter for each annotation per text_id
df_male['annotation_num'] = df_male.groupby('tweet_id_comment_nr').cumcount() + 1

# Step 2: Pivot the DataFrame so each annotation becomes a column
df_wide = df_male.pivot(index='tweet_id_comment_nr', columns='annotation_num', values='comment_respectvol_binary')

# Step 3: Reset index and rename columns (optional)
df_wide.columns = [f'annotation_{i}' for i in df_wide.columns]
df_wide = df_wide.reset_index()

# Step 4: (Optional) Merge the original text back in if needed
# Get one copy of each text
text_lookup = df[['tweet_id_comment_nr', 'text']].drop_duplicates(subset='tweet_id_comment_nr')
df_male_respect = df_wide.merge(text_lookup, on='tweet_id_comment_nr', how='left')

# Final result
df_male_respect
```

```python
# Count non-NaN annotations across annotation columns
annotation_cols = [col for col in df_male_respect.columns if col.startswith('annotation_')]

df_male_respect['num_valid_annotations'] = df_male_respect[annotation_cols].notna().sum(axis=1)

# Compute row-wise mode (majority vote)
def majority_vote(row):
    values = row[annotation_cols].dropna()
    if len(values) == 0:
        return np.nan
    return values.mode().iloc[0]  # mode() returns a Series

df_male_respect['majority_vote_respect'] = df_male_respect.apply(majority_vote, axis=1)
df_male_respect.head()
```

```python
df_male_respect['majority_vote_respect'].value_counts()
```

```python
df_male_respect_red = df_male_respect[["tweet_id_comment_nr","text",'majority_vote_respect']]
df_male_respect_red
```

### 3.2 Attack

```python
# Step 1: Add a counter for each annotation per text_id
df_male['annotation_num'] = df_male.groupby('tweet_id_comment_nr').cumcount() + 1

# Step 2: Pivot the DataFrame so each annotation becomes a column
df_wide = df_male.pivot(index='tweet_id_comment_nr', columns='annotation_num', values='comment_persoonlijke_aanval_binary')

# Step 3: Reset index and rename columns (optional)
df_wide.columns = [f'annotation_{i}' for i in df_wide.columns]
df_wide = df_wide.reset_index()

# Step 4: (Optional) Merge the original text back in if needed
# Get one copy of each text
text_lookup = df[['tweet_id_comment_nr', 'text']].drop_duplicates(subset='tweet_id_comment_nr')
df_male_attack = df_wide.merge(text_lookup, on='tweet_id_comment_nr', how='left')

# Compute row-wise mode (majority vote)
def majority_vote(row):
    values = row[annotation_cols].dropna()
    if len(values) == 0:
        return np.nan
    return values.mode().iloc[0]  # mode() returns a Series

df_male_attack['majority_vote_attack'] = df_male_attack.apply(majority_vote, axis=1)
df_male_attack.head()
```

```python
df_male_attack['majority_vote_attack'].value_counts()
```

```python
df_male_attack_red = df_male_attack[["tweet_id_comment_nr",'majority_vote_attack']]
df_male_attack_red
```

### 3.3 Argument

```python
# Step 1: Add a counter for each annotation per text_id
df_male['annotation_num'] = df_male.groupby('tweet_id_comment_nr').cumcount() + 1

# Step 2: Pivot the DataFrame so each annotation becomes a column
df_wide = df_male.pivot(index='tweet_id_comment_nr', columns='annotation_num', values='comment_inhoudelijk_argument_binary')

# Step 3: Reset index and rename columns (optional)
df_wide.columns = [f'annotation_{i}' for i in df_wide.columns]
df_wide = df_wide.reset_index()

# Step 4: (Optional) Merge the original text back in if needed
# Get one copy of each text
text_lookup = df[['tweet_id_comment_nr', 'text']].drop_duplicates(subset='tweet_id_comment_nr')
df_male_argument = df_wide.merge(text_lookup, on='tweet_id_comment_nr', how='left')

# Compute row-wise mode (majority vote)
def majority_vote(row):
    values = row[annotation_cols].dropna()
    if len(values) == 0:
        return np.nan
    return values.mode().iloc[0]  # mode() returns a Series

df_male_argument['majority_vote_argument'] = df_male_attack.apply(majority_vote, axis=1)
df_male_argument.head()
```

```python
df_male_argument['majority_vote_argument'].value_counts()
```

```python
df_male_argument_red = df_male_argument[["tweet_id_comment_nr",'majority_vote_argument']]
df_male_argument_red
```

### 3.4 Concat

```python
df_male_final = df_male_respect_red.merge(df_male_attack_red, on='tweet_id_comment_nr', how='outer') \
            .merge(df_male_argument_red, on='tweet_id_comment_nr', how='outer')
df_male_final
```

## 4. Short Text Cleansing Session

```python
# Get rid the @USER handle
df_male_final['text_cleaned_01'] = df_male_final['text'].str.replace(r'@USER', '', regex=True).str.strip()
```

```python
# Get rid of URLs 
df_male_final['text_cleaned_02'] = df_male_final['text_cleaned_01'].str.replace(r'http\S+', '', regex=True).str.strip()
```

```python
# Get rid of very short terms, such as ja, Dito., !! 
low_content_pattern = r'^\s*([a-zA-Z]{1,4}|[!?.,]{1,})\s*$'
df_male_final = df_male_final[~df_male_final['text_cleaned_02'].str.match(low_content_pattern, na=False)]
```

```python
# Remove now empty cells
df_male_final = df_male_final[df_male_final['text_cleaned_02'].str.strip().astype(bool)]
len(df_male_final)
```
