# Calculate and Compare Agreement/Reliability by Gender 

```python
import pandas as pd
```

```python
from sklearn.metrics import cohen_kappa_score
import krippendorff
```

## 1. Read Female and Male Datasets

```python
df_female_test = pd.read_csv("df_female_test.csv")
df_female_test
```

```python
df_male_test = pd.read_csv("df_male_test.csv")
df_male_test
```

## 2. Calculate Intercoder Agreement

### 2.1 Personal Attack

```python
# Calculate Cohen's Kappa
kappa = cohen_kappa_score(df_male_test['majority_vote_attack'], df_female_test['majority_vote_attack'])

print(f"Cohen's Kappa Score: {kappa}")
```

```python
data_matrix = df_male_test['majority_vote_attack'],df_female_test['majority_vote_attack'].to_numpy().T  # Transposing to get "item-by-coder" matrix format

# Calculate Krippendorff's alpha for nominal data
alpha_nominal = krippendorff.alpha(reliability_data=data_matrix, level_of_measurement="nominal")

print(f"Krippendorff's Alpha (Nominal): {alpha_nominal}")
```

### 2.2 Respect

```python
# Calculate Cohen's Kappa
kappa = cohen_kappa_score(df_male_test['majority_vote_respect'], df_female_test['majority_vote_respect'])

print(f"Cohen's Kappa Score: {kappa}")
```

```python
data_matrix = df_male_test['majority_vote_respect'],df_female_test['majority_vote_respect'].to_numpy().T  # Transposing to get "item-by-coder" matrix format

# Calculate Krippendorff's alpha for nominal data
alpha_nominal = krippendorff.alpha(reliability_data=data_matrix, level_of_measurement="nominal")

print(f"Krippendorff's Alpha (Nominal): {alpha_nominal}")
```

### 2.3 Substantial Argument

```python
# Calculate Cohen's Kappa
kappa = cohen_kappa_score(df_male_test['majority_vote_argument'], df_female_test['majority_vote_argument'])

print(f"Cohen's Kappa Score: {kappa}")
```

```python
data_matrix = df_male_test['majority_vote_argument'],df_female_test['majority_vote_argument'].to_numpy().T  # Transposing to get "item-by-coder" matrix format

# Calculate Krippendorff's alpha for nominal data
alpha_nominal = krippendorff.alpha(reliability_data=data_matrix, level_of_measurement="nominal")

print(f"Krippendorff's Alpha (Nominal): {alpha_nominal}")
```
