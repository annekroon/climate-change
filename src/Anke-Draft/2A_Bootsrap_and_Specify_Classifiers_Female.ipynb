## 2. Bootsstrap samples for few-shot learning

### 2.1 Respect

```python
df_female_train_respect = df_female_train[["tweet_id_comment_nr","text_cleaned_02",'majority_vote_respect']]
df_female_train_respect
```

✅ Draw 500 samples
✅ Each sample contains 10 texts, 5 with annotation == 1, 5 with annotation == 0
✅ Sampling is with replacement across samples
✅ Texts can appear multiple times across different samples (bootstrapping style)
✅ But each sample still contains 10 unique texts (no duplicates within a sample)

```python
# Step 1: Split the dataset by annotation 0/1
texts_1 = df_female_train_respect[df_female_train_respect['majority_vote_respect'] == 1]['text_cleaned_02'].unique()
texts_0 = df_female_train_respect[df_female_train_respect['majority_vote_respect'] == 0]['text_cleaned_02'].unique()


# Sanity check
assert len(texts_1) >= 5 and len(texts_0) >= 10, "Not enough texts in each class."

# Step 2: Draw 500 samples of 10 texts per class
samples_1 = []
samples_0 = []

for _ in range(500):
    sample_1 = np.random.choice(texts_1, size=10, replace=False)
    sample_0 = np.random.choice(texts_0, size=10, replace=False)
    samples_1.append(sample_1)
    samples_0.append(sample_0)

# Step 3: Convert to DataFrames
bootstrap_1_df = pd.DataFrame(samples_1, columns=[f'text_1_{i+1}' for i in range(10)])
bootstrap_0_df = pd.DataFrame(samples_0, columns=[f'text_0_{i+1}' for i in range(10)])

# ✅ Done!
print("Annotation = 1 Samples:", len(bootstrap_1_df))

print("\nAnnotation = 0 Samples:", len(bootstrap_0_df))
```

🔁 Generate 500 distinct few-shot prompts, where each one is built from:

One row from df_0 (10 examples of class 0)

One row from df_1 (10 examples of class 1)

💾 And save each prompt as a separate JSON file for use later during classification.

```python
import json
#import os

def generate_fewshot_sets(df_0, df_1, output_dir="fewshot_sets_female_respect"):
    os.makedirs(output_dir, exist_ok=True)

    for i in range(len(df_0)):
        fewshot = []

        # System message
        fewshot.append({
            "role": "system",
            "content": "You are a classification algorithm that has to differentiate between respectful and not respectful user comments in Dutch, where respectful is 1 and not respectful is 0. Answer with the correct class only. You must answer every time."
        })

        # Positive examples (label 1) from df_1
        for text in df_1.iloc[i]:
            fewshot.append({
                "role": "user",
                "content": f"{text} + categories: 1, 0"
            })
            fewshot.append({
                "role": "assistant",
                "content": "1"
            })

        # Negative examples (label 0) from df_0
        for text in df_0.iloc[i]:
            fewshot.append({
                "role": "user",
                "content": f"{text} + categories: 1, 0"
            })
            fewshot.append({
                "role": "assistant",
                "content": "0"
            })

        # Save to JSON file
        with open(os.path.join(output_dir, f"fewshot_{i}.json"), "w", encoding="utf-8") as f:
            json.dump(fewshot, f, ensure_ascii=False, indent=2)

    print(f"✅ Saved {len(df_0)} few-shot combinations to '{output_dir}/'")
```

### 3. Specifiy 500 Few-Shot Models

```python
from tqdm import tqdm

def prepare_fewshot_test_prompts_flat(fewshot_dir, test_comments, output_dir="models"):
    os.makedirs(output_dir, exist_ok=True)

    for i in tqdm(range(500), desc="Creating model files"):
        # Load few-shot base prompt
        with open(os.path.join(fewshot_dir, f"fewshot_{i}.json"), "r", encoding="utf-8") as f:
            base_prompt = json.load(f)

        # For each test comment, create a full prompt
        full_prompts = []
        for comment in test_comments:
            prompt = base_prompt + [{
                "role": "user",
                "content": f"{comment} + categories: 1, 0"
            }]
            full_prompts.append(prompt)

        # Save all test prompts for this model in one file
        with open(os.path.join(output_dir, f"model_{i:03d}.json"), "w", encoding="utf-8") as f:
            json.dump(full_prompts, f, ensure_ascii=False, indent=2)

    print(f"✅ Done! Saved 500 converted few-shot test sets in '{output_dir}/'")
```
